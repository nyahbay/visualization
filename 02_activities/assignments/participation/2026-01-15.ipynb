# Make basic figure

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100,50)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
font1 = {'family':'sans-serif', 'color':'blue', 'size':20}
font2 = {'family':'monospace', 'color':'green', 'size':14}
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x, y, marker = '*', color = "indigo")
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x, y, marker = '*', color = "indigo", linestyle = '--', linewidth = 2)
```

# Activity

```python
# Import libraries
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length)
plt.show()
```

```python
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length, cmap="Reds", fill=True)
plt.show()
```

```python
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length,
            cmap="Blues", fill=True, bw_adjust=0.5)
plt.show()
```

## Reflection

Aesthetic: The density plot appears polished and visually coherent. However, heavy smoothing can oversimplify the data and may obscure important structure.

Substantive: The plot is effective when the goal is to show where observations concentrate across two continuous variables. However, it hides individual data points and does not convey sample size. Outliers and multimodal distributions can also be obscured depending on the chosen kernel density bandwidth.

Perceptual: contours encode “more/less dense” in a way viewers can interpret quickly. However, viewers may interpret contour heights quantitatively without a colorbar/legend. And bandwidth changes the patterns. Different settings can lead to different conclusions.
