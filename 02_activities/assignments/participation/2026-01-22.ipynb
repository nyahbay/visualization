```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Set the random seed so results are reproducible
np.random.seed(613)

# Create an array of integers from 0 to 49 to use as x-values
x = np.arange(50)

# Generate 50 random integer y-values between 0 and 99
y = np.random.randint(0, 100,50)

# Create a figure, figsize controls size of output plot in inches
fig, ax = plt.subplots(figsize=(5, 3))

# Add horizontal grid lines to improve readability
ax.grid(axis = 'y')

# Plot y versus x with customized styling
ax.plot(x, y, 
        marker = '*',          # Symbol used for data points
        markersize = 12,       # Size of the marker
        color = '#7425b9',   # Line color
        linestyle = '--',      # Dashed line
        linewidth = 2,         # Thickness of the line
        markeredgecolor = '#fa9359',    # Color of marker outline
        markerfacecolor = '#000000')    # Fill color of marker
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)

x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
```

```python
# Create a figure
fig, ax = plt.subplots(figsize=(5, 3))

# Plot both y variables against the same x values
ax.plot(x, y1)
ax.plot(x, y2)
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1" )
ax.plot(x,y2, label = "Person 2" )
ax.legend(loc='lower right')
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")

ax.legend(loc='lower right',
          frameon = True, # add frame around the legend
          fontsize = 12, # change font size
          ncol = 2, # specify number of columns
          shadow = True )

fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")

ax.legend(loc='upper left',
          bbox_to_anchor =(1, 1))

fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")

ax.legend(loc='lower right')
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")

ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!")

fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")

ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!",
        ha='center', # alignment of text
        color = 'red', # modify font colour
        size = 20) # modify font size

fig.show()
```

```python
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
```

```python
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])

ax.text(1, 5, ".Data:(1, 5)", transform=ax.transData)
ax.text(0.5, 0.1, ".Axes:(0.5, 0.1)", transform=ax.transAxes)
ax.text(0.2, 0.2, ".Figure:(0.2, 0.2)", transform=fig.transFigure)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")

ax.legend(loc='lower right')

ax.annotate('This is important!', 
            xy=(10, 95), 
            xytext=(20, 94),
            arrowprops=dict(facecolor='black'))

fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")

ax.legend(loc='lower right')

ax.annotate('This is important!',
            xy=(10, 95), 
            xytext=(20, 94),
            arrowprops = dict(arrowstyle = "wedge", color = "hotpink"))

fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")

ax.legend(loc='lower right')

ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")

ax.legend(loc='lower right')

ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")

ax.legend(loc='lower right')

ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")

ax.legend(loc='lower right')
          
plt.xticks(rotation=45, ha='right')
```
