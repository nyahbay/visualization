### Describe the data visualization on this slide according to each of the four types of descriptive content.

Level 1 (chart elements): 
- A horizontal timeline chart titled “The Daily Routines of Famous Creative People.”
- Each row represents one person.
- The x-axis shows 24 hours (12 AM–12 PM–12 AM).
- Colored blocks show activities: sleep, creative work, day job/admin, food/leisure, exercise, other, with a legend at the top.
- Each person’s day is segmented into coloured bars across the timeline.

Level 2 (descriptive statistics): 
- Most individuals spend large continuous blocks sleeping (≈6–9 hours).
- Creative work commonly occurs in multi-hour stretches, often in the morning or late night.
- Many schedules include multiple work sessions separated by meals/leisure.
- Some people show more fragmented days, while others show regular, repeated patterns.
- Exercise appears in short, infrequent blocks for only some people.

Level 3 (complex trends):
- There is no single “ideal” routine.
- Creative work clusters either early morning or late night.
- Historical figures often mix day jobs with creative time, indicating creativity is scheduled around obligations.
- Some routines are highly structured and consistent, while others are irregular and segmented.
- Sleep timing varies widely, but is a dominant portion of each day.

Level 4 (context and domain insights):
- The visualization suggests that successful creative people follow diverse schedules, challenging the idea of one optimal daily routine.
- It implies creativity can be adapted to personal habits and constraints, not confined to standard work hours.
- The chart frames productivity as time allocation, encouraging reflection on how daily structure supports creative output.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])

fig,(ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))

ax1.scatter(x1,y1)
ax2.bar(x2,y2)
fig.show()
```

```python
fig,(ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))

ax1.scatter(x1, y1,
            marker="o",
            s=60,
            edgecolor="black",
            alpha=0.8)

ax1.set_title("Random Values", fontsize=13)
ax1.grid(True, linestyle="--", alpha=0.4)


bars = ax2.bar(x2, y2,
               hatch="//",      
               edgecolor="black")

ax2.set_title("Character Power Score", fontsize=13)
ax2.set_xlabel("Characters", fontsize=13)
ax2.set_ylabel("Power Score", fontsize=13)

plt.tight_layout()
plt.show()
```

```python
# first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))

# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "indigo")
ax.plot(x, y2,
color = "red")
```

```python
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")

ax.errorbar(x, #our x values
y2, #our y values
yerr = y2_sd,
fmt = "none")
```

```python
import requests
from io import BytesIO
from PIL import Image

url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"

headers = {"User-Agent": "Mozilla/5.0"}  # important
response = requests.get(url, headers=headers)

print(response.status_code)  # should be 200

image = Image.open(BytesIO(response.content))
image.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")

ax_image = fig.add_axes([0.1, # x coordinate (ON FIGURE, NOT AXES)
                         0.11, # y coordinate (ON FIGURE, NOT AXES)
                         0.15, # image width
                         0.35] # image height
)
```

```python
fig, ax = plt.subplots(figsize=(7, 3))

ax.plot(x, y2, color = "red")
ax_image = fig.add_axes([0.1, 0.11, 0.15, 0.35])
ax_image.imshow(image)
ax_image.axis('off')
```
