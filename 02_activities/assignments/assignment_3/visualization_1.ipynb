```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("Intimate Partner and Family Violence.csv")

# Aggregate total reported incidents by year
trend = (
    df.groupby("REPORT_YEAR", as_index=False)["COUNT_"]
      .sum()
      .sort_values("REPORT_YEAR")
)

# Define x and y for plotting
x = trend["REPORT_YEAR"]
y = trend["COUNT_"]

# Create figure
fig, ax = plt.subplots(figsize=(5, 3), dpi = 300)

# Plot the time series
ax.plot(
    x,
    y,
    marker='o',
    markersize=6,
    color='#A020F0',
    linestyle='-',
    linewidth=2.2,
    markeredgecolor='#A020F0',
    markerfacecolor='#eee1ff'
)

# Label axes and title
ax.set_xlabel("Year")
ax.set_ylabel("Number of Reports")
ax.set_title("Reported Intimate Partner and Family Violence Over Time")

# Add grid to improve readability
ax.grid(True, linestyle='--', linewidth=0.6, alpha=0.6)

# Adjust layout
plt.tight_layout()

# Display the plot
plt.show()
```
